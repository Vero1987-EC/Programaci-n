## Introducción al Encapsulamiento y acceso a miembros

El encapsulamiento es uno de los cuatro pilares fundamentales de la Programación Orientada a Objetos (POO). Se refiere a la restricción del acceso a ciertos componentes de un objeto, asegurando que solo se pueda acceder a ciertos atributos y métodos de un objeto desde fuera de la clase de la manera deseada.

### Pregunta problema: 
Imagina que estás diseñando una clase Autor para una biblioteca. ¿Cómo puedes garantizar que ciertos atributos, como el nombre, la nacionalidad o la fecha de nacimiento, no sean modificados accidentalmente o accedidos de manera inapropiada?

## Conceptos clave

_Encapsulamiento:_ Es la técnica de hacer que los campos de una clase sean privados o protegidos y proporcionar acceso a través de métodos públicos.

Diferencias entre "_" y "__" en Python:

_: Se utiliza para indicar que un atributo o método es protegido. Aunque técnicamente aún es accesible desde fuera de la clase, por convención se entiende que no debe ser accedido directamente. Sin embargo, esta protección es solo una convención y no impide realmente el acceso.

__: Se utiliza para indicar que un atributo o método es privado. Esto modifica el nombre del atributo/método de manera que es más difícil de acceder desde fuera de la clase, proporcionando una capa adicional de seguridad.

## Ejemplos prácticos

### Clase Autor con atributos públicos:

```python
class Autor:
    def __init__(self, nombre, nacionalidad):
        self.nombre = nombre
        self.nacionalidad = nacionalidad

    def mostrar_autor(self):
        print(f"Nombre: {self.nombre}")
        print(f"Nacionalidad: {self.nacionalidad}")
```

Este diseño, aunque funcional, no protege los atributos nombre y nacionalidad de modificaciones no deseadas.

### Clase Autor con atributos protegidos:

```python
class Autor:
    def __init__(self, nombre, nacionalidad):
        self._nombre = nombre
        self._nacionalidad = nacionalidad
```

Aquí, hemos marcado los atributos como protegidos usando un solo guion bajo. Esto indica que no deben ser accedidos directamente, pero aún es posible hacerlo.

### Clase Autor con atributos privados y métodos getter y setter:

```python
class Autor:
    def __init__(self, nombre, nacionalidad):
        self.__nombre = nombre
        self.__nacionalidad = nacionalidad

    def get_nombre(self):
        return self.__nombre

    def set_nombre(self, nombre):
        self.__nombre = nombre

    def get_nacionalidad(self):
        return self.__nacionalidad

    def set_nacionalidad(self, nacionalidad):
        self.__nacionalidad = nacionalidad
```

Con los atributos marcados como privados, ahora es mucho más difícil acceder a ellos desde fuera de la clase. Los métodos getter y setter nos permiten controlar cómo se accede y modifica la información.

## Encapsulamiento de métodos

El encapsulamiento no se limita solo a los atributos; también podemos encapsular métodos. Esto será explorado en profundidad en el tema de polimorfismo.

##  Encapsulamiento con Atributos Privados y Métodos de Acceso (Getters y Setters)

Al definir atributos con **doble guion bajo** (`self.__nombre`), Python aplica *Name Mangling*, ocultando la variable interna y bloqueando el acceso o modificación directa desde el exterior.

Para interactuar de manera segura con los datos encapsulados, implementamos la interfaz pública mediante **métodos de acceso**:

---

### 1. Métodos Getter (`get_nombre()`, `get_nacionalidad()`)
* **Propósito:** Permiten consultar el valor de un atributo privado.
* **Ventaja:** El código cliente puede leer la información sin poder modificar el estado interno del objeto por accidente.

---

### 2. Métodos Setter (`set_nombre()`, `set_nacionalidad()`)
* **Propósito:** Permiten modificar el valor de un atributo privado.
* **Ventaja:** Actúan como un filtro de seguridad donde se pueden incorporar **validaciones de negocio** (evitar cadenas vacías, números negativos, formatos incorrectos) antes de aplicar la modificación.

---

### 3. Encapsulamiento de Métodos (`def __metodo_privado(self):`)
* **Concepto:** Al igual que los atributos, los **métodos auxiliares internos** también pueden ser privados colocándoles doble guion bajo (`__`).
* **Utilidad:** Se usan para algoritmos o validaciones internas que no deben ser llamados directamente por el usuario de la clase, manteniendo limpia la interfaz pública del objeto.

 ### Desafío 52 

Atributos Privados (__titulo, __autor, __isbn): Se antepone el doble guion bajo (__) para activar el mecanismo de Name Mangling de Python, impidiendo que el código externo acceda o altere directamente los valores sin supervisión.

Getters (get_titulo(), get_autor(), get_isbn()): Permiten la consulta controlada de los atributos privados.

Setters (set_titulo(), set_autor(), set_isbn()): Reciben el nuevo valor y aplican validaciones (por ejemplo, verificar que la cadena no sea vacía o contenga solo espacios) antes de actualizar el estado del objeto.


### Desafío 53

Atributo Privado __libros: Se añade a la lista inicial del constructor self.__libros = [] con doble guion bajo para impedir que desde fuera se pueda vaciar, sobrescribir o alterar la colección directamente (por ejemplo, haciendo autor.__libros = []).

Método agregar_libro(self, libro): Define la interfaz pública autorizada para incorporar elementos a la colección privada, permitiendo realizar validaciones de duplicados antes de insertar.

Protección del Encapsulamiento en el Getter (get_libros()): Para evitar la fuga de referencias (encapsulation leak), el método get_libros() no devuelve la lista original directamente, sino una copia de la misma (list(self.__libros)). Así, si el código externo altera o borra la lista recibida, el estado interno del objeto Autor se mantiene intacto.

### Desafío 54

En Python, la forma más elegante y pythónica de implementar el encapsulamiento es mediante decoradores:
Decorador @property (Getter): Transforma un método de lectura en un atributo sintáctico. Permite consultar autor.nombre como si fuera una variable pública, pero ejecutando por detrás el método privado.

Decorador @<atributo>.setter (Setter): Intercepta la reasignación de valores (autor.nombre = "Nuevo Nombre"), permitiendo validar el dato antes de guardarlo en self.__nombre.

Ventajas sobre get_ / set_ explícitos:
Mantiene una sintaxis limpia y natural de acceso (objeto.atributo).
Permite agregar validaciones a posteriori sin romper el código cliente existente.
