## Situación problema

En el cuaderno anterior, vimos cómo implementar un TAD para una lista y discutimos algunas desventajas de ese enfoque. ¿Cómo podemos superar estas desventajas y crear una implementación más robusta, reutilizable y extensible de una lista? La respuesta está en el paradigma de objetos.

## Introducción al paradigma de objetos

El paradigma de objetos es un enfoque de programación que combina datos y operaciones en entidades llamadas 'objetos'. Estos objetos son instancias de 'clases', que actúan como plantillas. A través de la herencia, el polimorfismo y el encapsulamiento, el paradigma de objetos nos permite crear programas más estructurados y modulares.

En el siguiente cuaderno, exploraremos en detalle cómo implementar clases y objetos en Python, y cómo estos conceptos pueden ayudarnos a superar las desventajas del enfoque TAD que vimos anteriormente.

# Paradigma de Objetos: Conceptos Básicos y Principios

## 1. Situación Problema
En la unidad anterior analizamos el **TAD Lista** implementado de forma procedural (usando una variable global `lista` y funciones independientes). Aunque funcional para casos simples, este enfoque presentó serias limitaciones:
* **Vulnerabilidad de los datos:** Al no haber ocultamiento de información, cualquier parte del programa puede alterar la lista directamente.
* **Incapacidad de escalar:** Imposibilidad de gestionar múltiples listas independientes de forma limpia.

**Solución:** El **Paradigma de Programación Orientada a Objetos (POO)** resuelve estas falencias integrando datos y comportamientos dentro de una misma estructura.

---

## 2. Introducción al Paradigma de Objetos

La **POO** es un modelo de programación que organiza el código en torno a **objetos** en lugar de funciones y lógica pura.

### Conceptos Fundamentales:
* **Clase:** Es el plano o molde abstracto que define las propiedades y comportamientos que tendrán los objetos.
* **Objeto:** Es una instancia concreta de una clase que posee un estado propio y puede ejecutar métodos.
* **Encapsulamiento:** Oculta los detalles internos de los datos y expone únicamente una interfaz segura.
* **Herencia:** Permite reutilizar y extender código creando nuevas clases a partir de clases existentes.
* **Polimorfismo:** Permite que diferentes clases respondan al mismo mensaje o método según su propia implementación.
