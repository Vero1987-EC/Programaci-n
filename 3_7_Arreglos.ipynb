# Añadir elementos a un array en Python

En Python, puedes añadir elementos a una lista usando el método append(). Este método añade el elemento al final de la lista.

# Eliminar elementos del array con pop ( ) o remove ( )

Para eliminar elementos de una lista, puedes usar el método pop() o remove(). pop() elimina el elemento en la posición especificada, mientras que remove() elimina el primer elemento con el valor especificado.

```python
productos.pop(1)  # Elimina el elemento en el índice 1
productos.remove("cerezas")  # Elimina la primera aparición de "cerezas"
print(productos)
```

## 4. Métodos Principales para Manejo de Arreglos (Listas) en Python

Las listas en Python incluyen métodos integrados para agregar, eliminar, buscar y ordenar elementos:

| Método | Descripción | Ejemplo |
| :--- | :--- | :--- |
| **`append(x)`** | Agrega el elemento `x` al final de la lista. | `lista.append(5)` |
| **`insert(i, x)`** | Inserta el elemento `x` en la posición del índice `i`. | `lista.insert(0, "inicio")` |
| **`extend(iterable)`** | Agrega todos los elementos de otra lista al final. | `lista.extend([1, 2])` |
| **`pop(i)`** | Elimina y devuelve el elemento en el índice `i` (por defecto el último). | `ultimo = lista.pop()` |
| **`remove(x)`** | Elimina la primera aparición del valor `x`. | `lista.remove("manzana")` |
| **`clear()`** | Elimina todos los elementos de la lista dejándola vacía. | `lista.clear()` |
| **`index(x)`** | Devuelve el índice numérico de la primera aparición del valor `x`. | `pos = lista.index("peras")` |
| **`count(x)`** | Cuenta cuántas veces se repite el valor `x` en la lista. | `cant = lista.count(1)` |
| **`sort()`** | Ordena los elementos de la lista de menor a mayor (in-place). | `lista.sort()` |
| **`reverse()`** | Invierte el orden actual de los elementos en la lista. | `lista.reverse()` |
| **`copy()`** | Devuelve una copia independiente de la lista. | `nueva = lista.copy()` |

```python

print("=== 1. CREACIÓN Y ACCESO POR ÍNDICE ===")
productos = ["manzanas", "bananas", "cerezas", "duraznos", "peras"]
primer_producto = productos[0]
print("Lista completa de productos:", productos)
print("El primer producto es:", primer_producto)


print("\n=== 2. AGREGAR Y ELIMINAR ELEMENTOS ===")
productos_sec2 = ["manzanas", "bananas", "cerezas", "duraznos", "peras"]
productos_sec2.append("kiwis")
print("Después de agregar 'kiwis':", productos_sec2)

# Eliminar por índice (índice 1 -> elimina 'bananas')
productos_sec2.pop(1)

# Eliminar por valor
productos_sec2.remove("cerezas")
print("Lista final luego de eliminar:", productos_sec2)


print("\n=== 3. INSPECCIÓN DE MEMORIA CON id() ===")
numeros = [1, 2, 3, 4, 5, "d", 6, 8, 9, 12, 24, 24, 1, 1, 1, 1]

print("--- Direcciones de memoria de cada elemento ---")
for i in range(len(numeros)):
  print(f"Posición {i} (Valor: {numeros[i]}): id = {id(numeros[i])}")

print("\n--- Diferencia de ID entre elementos contiguos ---")
for i in range(len(numeros) - 1):
  diferencia_id = id(numeros[i + 1]) - id(numeros[i])
  print(
      f"Elemento {numeros[i+1]} vs {numeros[i]} -> Diferencia de ID:"
      f" {diferencia_id}"
  )


print("\n=== 4. REUTILIZACIÓN DE MEMORIA Y COMPARACIÓN CON 'is' ===")
mi_lista = [1, 2, 3]  # Asignación correcta de la lista
print("ID de mi_lista:", id(mi_lista))

a = "z "
b = "z "
print("¿a y b comparten la misma memoria?:", a is b)
print("ID de a:", id(a))
print("ID de b:", id(b))


print("\n=== 5. DEMOSTRACIÓN DE MÉTODOS DE LISTAS ===")
frutas = ["manzanas", "bananas", "duraznos"]
print("Lista inicial de frutas:", frutas)

# Insertar y Extender
frutas.append("kiwis")
frutas.insert(1, "peras")
print("Tras append e insert:", frutas)

mas_frutas = ["uvas", "sandías"]
frutas.extend(mas_frutas)
print("Tras extend:", frutas)

# Buscar y Contar
posicion = frutas.index("peras")
print(f"La fruta 'peras' está en el índice: {posicion}")

frutas.append("manzanas")
print("Cantidad de 'manzanas':", frutas.count("manzanas"))

# Ordenar e Invertir
frutas.sort()
print("Lista ordenada alfabéticamente:", frutas)

frutas.reverse()
print("Lista invertida:", frutas)

# Eliminar y Copiar
eliminado = frutas.pop()  # Elimina 'bananas' que quedó al final
print(f"Elemento eliminado con pop(): {eliminado}")

frutas.remove("duraznos")  # Elimina 'duraznos'
print("Lista tras remove('duraznos'):", frutas)

copia_frutas = frutas.copy()
frutas.clear()
print("Lista original tras clear():", frutas)
print("Copia preservada en otra variable:", copia_frutas)
```

## Desafío 19: Ordenar el Inventario de Libros

**Consigna:**  
Como encargado de la biblioteca, necesitas organizar los libros de acuerdo con sus códigos de identificación en **orden decreciente**, sin modificar la lista original. Se recomienda usar la función `sorted()`.

#### Respuestas Teóricas del Desafío:

1. **¿Por qué es importante no modificar la lista original?**
   * En sistemas reales (como el inventario de una biblioteca), mantener la lista original preserva el **orden de ingreso histórico** o el registro primario de los datos. Si modificamos la lista original, se pierde la secuencia inicial y se pueden romper referencias externas.

2. **¿Por qué no puedo usar el método `.sort()` sobre la lista original?**
   * El método `lista.sort()` realiza un ordenamiento *in-place* (modifica la lista directamente y devuelve `None`).
   * La función integrada `sorted(lista)`, en cambio, **crea y devuelve una nueva lista ordenada**, dejando la lista original completamente intacta.

```python
# ==============================================================================
# DESAFÍO 19: ORDENAR EL INVENTARIO DE LIBROS
# ==============================================================================

# Lista original de códigos de identificación de libros en la biblioteca
codigos_libros = [1001, 1002, 1003, 1004, 1005]

print("--- ESTADO INICIAL DEL INVENTARIO ---")
print("Códigos originales:", codigos_libros)

# Ordenamos en orden decreciente usando sorted() con reverse=True
# Esto crea una nueva lista sin alterar la original
codigos_ordenados_decreciente = sorted(codigos_libros, reverse=True)

print("\n--- INVENTARIO ORGANIZADO (DECRECIENTE) ---")
print("Códigos ordenados decreciente:", codigos_ordenados_decreciente)

print("\n--- VERIFICACIÓN DE INTEGRIDAD DE LA LISTA ORIGINAL ---")
print("Lista original (intacta):", codigos_libros)
```

```python
# ==============================================================================
# DESAFÍO 20: CARACTERES ASCII
# ==============================================================================

# Lista de códigos numéricos ASCII
codigos_ascii = [72, 111, 108, 97, 32, 77, 117, 110, 100, 111] 

print("--- DESAFÍO 20: CONVERSIÓN A CARACTERES ASCII ---")
print("Lista de códigos numéricos:", codigos_ascii)

# Conversión usando comprensión de listas y chr()
caracteres = [chr(codigo) for codigo in codigos_ascii]
palabra = "".join(caracteres)

print("\nCaracteres individuales:", caracteres)
print(f"Palabra descifrada: {palabra}")
```

### Problema 1: Análisis de Ventas Mensuales usando el módulo array

**Consigna:**  
Como gerente de una tienda de ropa, necesitas analizar las ventas mensuales de un producto específico durante el último año utilizando el módulo `array`. El objetivo es calcular la venta total anual, el promedio de ventas mensuales, y determinar los meses con la venta más baja y más alta.

```python
import array

# ==============================================================================
# PROBLEMA 1: ANÁLISIS DE VENTAS MENSUALES (MÓDULO ARRAY)
# ==============================================================================

# 1. Datos iniciales: ventas mensuales en unidades (tipo 'i' para enteros)
ventas_mensuales = array.array(
    'i', [120, 135, 150, 140, 160, 155, 170, 165, 140, 130, 125, 180]
)

print("--- DATOS DE VENTAS MENSUALES ---")
print("Ventas por mes:", list(ventas_mensuales))

# 2. Venta total anual
total_anual = sum(ventas_mensuales)
print(f"\n1. Venta total anual: {total_anual} unidades")

# 3. Promedio de ventas mensuales
promedio_mensual = total_anual / len(ventas_mensuales)
print(f"2. Promedio de ventas mensuales: {promedio_mensual:.2f} unidades")

# 4. Meses con la venta más baja y más alta (+1 porque los meses van de 1 a 12)
mes_baja = ventas_mensuales.index(min(ventas_mensuales)) + 1
mes_alta = ventas_mensuales.index(max(ventas_mensuales)) + 1

print(
    f"3. Mes con la venta más baja: Mes {mes_baja} (Venta:"
    f" {min(ventas_mensuales)} unidades)"
)
print(
    f"   Mes con la venta más alta: Mes {mes_alta} (Venta:"
    f" {max(ventas_mensuales)} unidades)"
)
```

# Problema 2: Análisis de Temperaturas Diarias usando NumPy
Como meteorólogo, necesitas analizar las temperaturas diarias registradas en tu ciudad durante el último mes. Usarás la librería NumPy para trabajar eficientemente con los datos de temperaturas. El objetivo es calcular la temperatura media, la temperatura máxima y mínima, y visualizar las temperaturas en un gráfico de línea.

## Datos Iniciales
Las temperaturas diarias registradas en el último mes (30 días) son las siguientes:

```python
import numpy as np

temperaturas_diarias = np.array([22.5, 21.3, 23.1, 25.0, 24.5, 22.1, 23.7, 24.8, 25.3, 26.1, 
                                 23.4, 24.0, 22.9, 21.5, 23.0, 24.3, 25.0, 24.7, 23.1, 22.4, 
                                 23.5, 24.1, 25.4, 26.0, 24.8, 23.9, 22.7, 23.3, 24.5, 25.1])
```

## Tareas:
**Calcula la temperatura media del mes.**

```python
temperatura_media = np.mean(temperaturas_diarias)
print(f"Temperatura media del mes: {temperatura_media:.2f}°C")
```

**Encuentra la temperatura máxima y mínima del mes.**

```python
temperatura_maxima = np.max(temperaturas_diarias)
temperatura_minima = np.min(temperaturas_diarias)
print(f"Temperatura máxima del mes: {temperatura_maxima}°C")
print(f"Temperatura mínima del mes: {temperatura_minima}°C")
```

**Genera un gráfico de línea de las temperaturas diarias.**

```python
import matplotlib.pyplot as plt

dias = np.arange(1, 31)
plt.plot(dias, temperaturas_diarias, marker='o')
plt.xlabel('Día')
plt.ylabel('Temperatura (°C)')
plt.title('Temperaturas Diarias del Último Mes')
plt.grid(True)
plt.show()
```

Como meteorólogo, necesitas analizar las temperaturas diarias registradas en tu ciudad durante el último mes (30 días). Usarás la librería **NumPy** para trabajar eficientemente con los datos y **Matplotlib** para visualizar las temperaturas en un gráfico de línea.

```python
# ==============================================================================
# PROBLEMA 2: ANÁLISIS DE TEMPERATURAS DIARIAS (NUMPY)
# ==============================================================================

import matplotlib.pyplot as plt
import numpy as np

# 1. Datos iniciales: Temperaturas registradas durante 30 días
temperaturas_diarias = np.array([
    22.5,
    21.3,
    23.1,
    25.0,
    24.5,
    22.1,
    23.7,
    24.8,
    25.3,
    26.1,
    23.4,
    24.0,
    22.9,
    21.5,
    23.0,
    24.3,
    25.0,
    24.7,
    23.1,
    22.4,
    23.5,
    24.1,
    25.4,
    26.0,
    24.8,
    23.9,
    22.7,
    23.3,
    24.5,
    25.1,
])

# 2. Tarea 1: Temperatura media del mes
temperatura_media = np.mean(temperaturas_diarias)
print(f"Temperatura media del mes: {temperatura_media:.2f}°C")

# 3. Tarea 2: Temperatura máxima y mínima del mes
temperatura_maxima = np.max(temperaturas_diarias)
temperatura_minima = np.min(temperaturas_diarias)
print(f"Temperatura máxima del mes: {temperatura_maxima}°C")
print(f"Temperatura mínima del mes: {temperatura_minima}°C")

# 4. Tarea 3: Gráfico de línea de las temperaturas diarias
dias = np.arange(1, 31)
plt.plot(dias, temperaturas_diarias, marker="o")
plt.xlabel("Día")
plt.ylabel("Temperatura (°C)")
plt.title("Temperaturas Diarias del Último Mes")
plt.grid(True)
plt.show()
```
